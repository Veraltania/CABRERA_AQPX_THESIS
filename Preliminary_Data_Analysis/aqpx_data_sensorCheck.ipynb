```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
# Load CSV
date = "2025-12-17"
df = pd.read_csv(f"AQPX_data_log_{date}.csv")

# Clean Time column
df["Time"] = df["Time"].astype(str).str.strip()

# Combine date + time into a proper datetime
dtime = pd.to_datetime(
    date + " " + df["Time"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)
```

```python
ec = df["MCP_WQ_EC"]
ph = df["MCP_WQ_PH"]
temp = df["MCP_WQ_TEMP"]
turb = df["MCP_WQ_TURB"]

#downscale values
factor = 20
dtime_small = dtime[::factor]
ec_small = ec[::factor]
ph_small = ph[::factor]
temp_small = temp[::factor]
turb_small = turb[::factor]
```

```python
plt.figure(figsize=(12,4))
plt.plot(dtime_small, ec_small)

plt.xlabel("Time")
plt.ylabel("EC (MCP_WQ_EC")

# HOURLY TICKS
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=1))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(dtime_small, ph_small)

plt.xlabel("Time")
plt.ylabel("pH (MCP_WQ_PH")

# HOURLY TICKS
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=1))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(dtime_small, temp_small)

plt.xlabel("Time")
plt.ylabel("TEMPERATURE (MCP_WQ_TEMP")

# HOURLY TICKS
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=1))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(dtime_small, turb_small)

plt.xlabel("Time")
plt.ylabel("TURBIDITY (MCP_WQ_TURB")

# HOURLY TICKS
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=1))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
