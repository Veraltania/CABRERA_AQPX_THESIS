```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
# Load CSV
df = pd.read_csv("AQPX_data_log_2026-01-08.csv")

dtds = df["MCP_WQ_EC"]
dtime = pd.to_datetime("2025-12-16 " + df["Time"], errors="coerce")
dtime_orig = dtime
```

```python
#downscale values
factor = 60
dtime_small = dtime[::factor]
dtds_small = dtds[::factor]
```

```python
plt.figure(figsize=(12,4))
plt.plot(dtime_small, dtds_small)

plt.xlabel("Time")
plt.ylabel("Electrical Conductivity (MCP_WQ_EC)")
plt.title("Electrical Conductivity vs Time")

# HOURLY TICKS
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=1))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
